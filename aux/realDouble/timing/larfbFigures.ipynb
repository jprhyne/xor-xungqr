```python
# Package imports
import numpy as np

# My own files
from aux import *
from parse import *
from perf import *
```

```python
(mVec, nVec, kVec, \
            opt_time, ref_time, new_time, new_solve_time) = parseLarfbFile("perf_files/rect_larfb_12_08_25.txt")
```

```python
xVec = kVec
yVecs = np.array([opt_time, ref_time, new_time, new_solve_time])
for i in range(yVecs.shape[0]):
    yVecs[i][:] = larfbPerf(mVec,nVec,kVec,yVecs[i][:])
labelVecs = np.array(["AOCL", "Reference", "DLARFB0C2", "DLARFB0C2_SOLVE"])
xLabel = "Number of Columns in C (n)"
yLabel = "Performance: GFlop/s"
title = r'GFlop/s for dlarfb with $m=10000, n=k$ varying'
createFigure(xVec, yVecs, labelVecs, title, xLabel, yLabel, filename="figures/rect_larfb_10000.png")
```
